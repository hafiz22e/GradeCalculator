```python
import tkinter as tk
from tkinter import messagebox

class GradeCalculator:
    def __init__(self, root):
        self.root = root
        self.root.title("Grade & Negative Marking Calculator")
        self.root.geometry("400x500")

        # Configuration variables
        self.total_marks = 0
        self.total_questions = 0
        self.marks_per_q = 0

        self.setup_ui()

    def setup_ui(self):
        """Initial screen to set total marks and questions."""
        self.clear_screen()
        
        tk.Label(self.root, text="--- Initial Setup ---", font=('Arial', 12, 'bold')).pack(pady=10)
        
        tk.Label(self.root, text="Enter Total Marks:").pack()
        self.ent_total_marks = tk.Entry(self.root)
        self.ent_total_marks.pack(pady=5)
        self.ent_total_marks.insert(0, "40")

        tk.Label(self.root, text="Enter Total Number of Questions:").pack()
        self.ent_total_qs = tk.Entry(self.root)
        self.ent_total_qs.pack(pady=5)
        self.ent_total_qs.insert(0, "20")

        tk.Button(self.root, text="Save & Start", command=self.save_config, bg="blue", fg="white").pack(pady=20)

    def save_config(self):
        try:
            self.total_marks = float(self.ent_total_marks.get())
            self.total_questions = int(self.ent_total_qs.get())
            self.marks_per_q = self.total_marks / self.total_questions
            self.calculation_ui()
        except Exception as e:
            messagebox.showerror("Error", "Please enter valid numbers.")

    def calculation_ui(self):
        """The main grading screen."""
        self.clear_screen()

        tk.Label(self.root, text=f"Total: {self.total_marks} Marks | {self.total_questions} Qs", font=('Arial', 10, 'italic')).pack(pady=5)
        tk.Label(self.root, text=f"Each Question = {self.marks_per_q} Marks", font=('Arial', 10)).pack()
        
        tk.Label(self.root, text="\nNumber of Correct Questions:").pack()
        self.ent_correct = tk.Entry(self.root)
        self.ent_correct.pack()

        tk.Label(self.root, text="Wrong Attempted:").pack()
        self.ent_wrong = tk.Entry(self.root)
        self.ent_wrong.pack()

        tk.Label(self.root, text="Not Attempted (Optional):").pack()
        self.ent_not_attempted = tk.Entry(self.root)
        self.ent_not_attempted.pack()

        tk.Button(self.root, text="Calculate Grade", command=self.calculate, bg="green", fg="white").pack(pady=15)
        
        self.res_label = tk.Label(self.root, text="", font=('Arial', 11, 'bold'), fg="blue")
        self.res_label.pack(pady=10)

        tk.Label(self.root, text="Type 'OK' or 'ok' to Exit:").pack(side="bottom")
        self.ent_exit = tk.Entry(self.root)
        self.ent_exit.pack(side="bottom", pady=5)
        self.ent_exit.bind("<Return>", self.check_exit)

    def calculate(self):
        try:
            correct = int(self.ent_correct.get() or 0)
            wrong = int(self.ent_wrong.get() or 0)
            
            if (correct + wrong) > self.total_questions:
                messagebox.showwarning("Warning", "Sum of questions exceeds total questions!")
                return

            # Logic: Correct marks - (Wrong count * 1)
            # We don't deduct anything for 'not attempted'
            raw_correct_marks = correct * self.marks_per_q
            final_marks = raw_correct_marks - (wrong * 1)

            result_text = f"Total marks {int(raw_correct_marks)} - {wrong} = {int(final_marks)} out of {int(self.total_marks)}"
            self.res_label.config(text=result_text)
            
            # Clear entries for next calculation automatically
            self.ent_correct.delete(0, tk.END)
            self.ent_wrong.delete(0, tk.END)
            self.ent_not_attempted.delete(0, tk.END)
            
        except ValueError:
            messagebox.showerror("Error", "Please enter valid integers for questions.")

    def check_exit(self, event):
        val = self.ent_exit.get().strip()
        if val.lower() == "ok":
            self.root.destroy()
        else:
            self.ent_exit.delete(0, tk.END)

    def clear_screen(self):
        for widget in self.root.winfo_children():
            widget.destroy()

if __name__ == "__main__":
    root = tk.Tk()
    app = GradeCalculator(root)
    root.mainloop()
```
